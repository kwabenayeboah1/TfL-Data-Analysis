## Accident Availability Checks

This script is a way of checking whether new years have been added to the APIby TfL, which will allow us to expand the accident data we're looking at.

At the moment, TfL only has data spanning the years 2005-2019 as shown below.

Running this script periodically would allow us to understand if further datasets have been added that can be further used in Data Analysis.

```python
import requests

valid_years=[]

for year in range(1990,2030):
    url = f"https://api.tfl.gov.uk/AccidentStats/{year}"
    r = requests.get(url)

    if r.status_code == 200 and r.json():
        valid_years.append(year)
    
min(valid_years),max(valid_years)
```

* Additional script to download the relevant years in one go.

```python
import json

BASE_URL = "https://api.tfl.gov.uk/AccidentStats/"
START_YEAR = 2005
END_YEAR = 2019

for year in range (START_YEAR,END_YEAR + 1):
    url = f"{BASE_URL}{year}"
    print(f"Fetching {url}")

    response = requests.get(url, timeout=30)

    if response.status_code != 200:
        print(f" Failed for {year}: {response.status_code}")
        continue

    data = response.json()
    
    filename = f"response_{year}.json"
    with open (filename, "w", encoding = "utf-8") as f:
        json.dump(data, f, ensure_ascii = False, indent=2)

    print(f"Saved {filename}")
```
